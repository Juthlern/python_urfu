### Задание 1
- Создание объект Series из словаря, где ключи - имена городов, а значения - населения этих городов. Выведите этот Series.
- Создайте DataFrame из двух списков одинаковой длины, преобразовав его к DataFrame и
добавив собственные названия столбцов. Посмотрите на индексы и названия столбцов.
- Используя созданный DataFrame , выведите его методами df.info(), df.shape и
df.describe()

````python
import pandas as pd

# --- Задание 1.1: Series из словаря, где ключи — города, значения — население ---
city_population = {
    'Москва': 12655050,
    'Санкт-Петербург': 5383890,
    'Новосибирск': 1625631,
    'Екатеринбург': 1493749,
    'Казань': 1300148
}
city_series = pd.Series(city_population)
print('1.1 Series из словаря:')
print(city_series)
print()

# --- Задание 1.2: DataFrame из двух списков одной длины ---
cities = ['Москва', 'Санкт-Петербург', 'Новосибирск', 'Екатеринбург', 'Казань']
populations = [12655050, 5383890, 1625631, 1493749, 1300148]
df = pd.DataFrame({
    'Город': cities,
    'Население': populations
})
print('1.2 DataFrame с городами и населением:')
print(df)
print('Индексы:', df.index)
print('Столбцы:', df.columns)
print()

# --- Задание 1.3: Методы df.info(), df.shape, df.describe() ---
print('1.3 Информация о DataFrame:')
print(df.info())
print('Форма таблицы (shape):', df.shape)
print('Статистика по числовым столбцам:')
print(df.describe())
````

````python
import pandas as pd
import json
import sqlite3

# --- 2.1 Сохраняем CSV, читаем DataFrame, выводим 3 строки ---
csv_data = """Name,Score,Passed
Alice,85,True
Bob,90,False
Charlie,78,True
Soul,12,False
David,100,True
Lisa,66,False
"""
with open('data.csv', 'w', encoding='utf-8') as f:
    f.write(csv_data)

df2 = pd.read_csv('data.csv')
print('2.1 Первые 3 строки DataFrame:')
print(df2.head(3))
print()

# --- 2.2 Создаём JSON-файл, читаем его с помощью pandas ---
products = [
    {"name": "Молоко", "price": 65},
    {"name": "Хлеб", "price": 30},
    {"name": "Яблоки", "price": 120}
]
with open('products.json', 'w', encoding='utf-8') as f:
    json.dump(products, f, ensure_ascii=False)

products_df = pd.read_json('products.json')
print('2.2 Содержимое JSON-файла:')
print(products_df)
print()

# --- 2.3 SQLite база, запись и чтение через pandas ---
conn = sqlite3.connect('test.db')
cur = conn.cursor()
cur.execute('CREATE TABLE IF NOT EXISTS students (id INTEGER PRIMARY KEY, name TEXT, score INTEGER)')
cur.execute('DELETE FROM students')
cur.executemany('INSERT INTO students (name, score) VALUES (?, ?)', [
    ('Alice', 85), ('Bob', 90), ('Charlie', 78)
])
conn.commit()

students_df = pd.read_sql('SELECT * FROM students', conn)
print('2.3 Данные из SQLite:')
print(students_df)
print()
conn.close()

# --- 2.4 Чтение HTML-таблицы с сайта ---
url = 'https://www.worldometers.info/world-population/population-by-country/'
tables = pd.read_html(url)
print('2.4 Содержимое HTML-таблицы (первые 3 строки):')
print(tables[0].head(3))
print()
````

- Отфильтруйте df из примера по условию: Age > 30 и Score >= 90.
- Создайте DataFrame с покупками (столбцы: User , Item , Quantity ), затем посчитайте общее
количество покупок по каждому User с помощью groupby().
- Отсортируйте один из ваших DataFrame по нескольким столбцам (например, сначала по
Department , потом по Salary ).
- Используя groupby() , найдите средний возраст студентов в каждом классе (придумайте данные с
классами).

````python
import pandas as pd

# --- 3.1 Фильтрация по Age > 30 и Score >= 90 ---
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [25, 30, 35, 40, 22],
    'Score': [88, 92, 85, 90, 95]
})
filtered = df[(df['Age'] > 30) & (df['Score'] >= 90)]
print('3.1 Отфильтрованные студенты (Age > 30 и Score >= 90):')
print(filtered)
print()

# --- 3.2 DataFrame с покупками, groupby по User ---
df_purchases = pd.DataFrame({
    'User': ['Ivan', 'Anna', 'Ivan', 'Petr', 'Anna', 'Ivan'],
    'Item': ['Bread', 'Milk', 'Eggs', 'Juice', 'Butter', 'Milk'],
    'Quantity': [1, 2, 10, 1, 2, 2]
})
purchases_by_user = df_purchases.groupby('User')['Quantity'].sum()
print('3.2 Сумма покупок по каждому пользователю:')
print(purchases_by_user)
print()

# --- 3.3 Сортировка DataFrame по Dept и Salary ---
df2 = pd.DataFrame({
    'Dept': ['HR', 'HR', 'IT', 'IT', 'HR'],
    'Salary': [100, 150, 200, 250, 180]
})
sorted_df2 = df2.sort_values(by=['Dept', 'Salary'], ascending=[True, False])
print('3.3 DataFrame, отсортированный по отделу (Dept) и зарплате (Salary, по убыванию):')
print(sorted_df2)
print()

# --- 3.4 Средний возраст по классам с groupby ---
df_students = pd.DataFrame({
    'Name': ['Oleg', 'Masha', 'Igor', 'Olga', 'Ivan', 'Nina'],
    'Class': ['10A', '10B', '10A', '10B', '10C', '10A'],
    'Age': [16, 17, 16, 17, 16, 16]
})
mean_age_by_class = df_students.groupby('Class')['Age'].mean()
print('3.4 Средний возраст по классам:')
print(mean_age_by_class)
print()
````

### Задание 4

- В вашем DataFrame найдите количество пропущенных значений в каждом столбце (используйте
df.isna().sum()).
- Удалите все строки, где присутствуют пропуски во всех столбцах одновременно.
- Создайте столбец с датами в строковом формате (например, `df['date_str'] =
['2023-01-01', ...]` ), затем преобразуйте его в datetime и извлеките из него год и месяц.
- Приведите числовой столбец с плавающей точкой к целочисленному типу, а строковый столбец с
числами – к числовому типу с помощью pd.to_numeric()

````python
import pandas as pd
import numpy as np

# --- 4.1 Количество пропущенных значений в каждом столбце ---
df = pd.DataFrame({
    'A': [1, 2, np.nan, 4],
    'B': [5, np.nan, np.nan, 8],
    'C': ['10', np.nan, '30', '40']
})
print('4.1 Количество пропущенных значений:')
print(df.isna().sum())
print()

# --- 4.2 Удалить все строки, где пропуски во всех столбцах одновременно ---
df_no_all_nan = df.dropna(how='all')
print('4.2 DataFrame без строк, полностью состоящих из пропусков:')
print(df_no_all_nan)
print()

# --- 4.3 Добавить столбец с датами, перевести к datetime, извлечь год и месяц ---
df_no_all_nan['date_str'] = ['2023-01-01', '2022-07-14', '2021-11-20', '2020-07-20']
df_no_all_nan['date'] = pd.to_datetime(df_no_all_nan['date_str'])
df_no_all_nan['year'] = df_no_all_nan['date'].dt.year
df_no_all_nan['month'] = df_no_all_nan['date'].dt.month
print('4.3 DataFrame с датами, годом и месяцем:')
print(df_no_all_nan[['date_str', 'date', 'year', 'month']])
print()

# --- 4.4 Привести числовой float к int, строку с числами — к float ---
df_no_all_nan['A_int'] = pd.to_numeric(df_no_all_nan['A'], downcast='integer')
df_no_all_nan['C_num'] = pd.to_numeric(df_no_all_nan['C'], errors='coerce')
print('4.4 Преобразованные типы:')
print(df_no_all_nan[['A', 'A_int', 'C', 'C_num']])
print()
````
